# Getting a requests session authenticated via Selenium

This is a little sample notebook for creating a requests session capable of requesting pages using cookies generated in a Selenium webdriver environment.

it needs selenium (use pip install selenium or conda install selenium)
and geckodriver, download from  https://github.com/mozilla/geckodriver/releases

```python
import requests
from selenium import webdriver
from bs4 import BeautifulSoup
import html5lib
```

```python
# Get the url for the page you want to log in on
#change path to match where geckodriver is installed
driver = webdriver.Chrome(executable_path=r'chromedriver')  # update this to your local chromedriver path, or leave as-is if chromedriver is on your PATH 
driver.get("http://pdharvard.egreenapple.com.ezp-prod1.hul.harvard.edu/index2.html")
```

## Log in to the page in the window that pops open before running the next cell!

```python
# Start a session in requests
s = requests.session()

# Set cookies in requests session from selenium session
for cookie in driver.get_cookies():
    c = {cookie['name']: cookie['value']}
    s.cookies.update(c)

# Show the new session cookies
s.cookies.get_dict()
```

```python
# Set session headers to look like Selenium session and add the referer url
user_agent = driver.execute_script("return navigator.userAgent;")
headers = {"User-Agent": user_agent, "Referer": "http://pdharvard.egreenapple.com.ezp-prod1.hul.harvard.edu/index2.html"}
s.headers.update(headers)

# Show the new headers
s.headers
```

```python
# Have requests get the main frame
response = s.get("http://pdharvard.egreenapple.com.ezp-prod1.hul.harvard.edu/advancesearch?channelid=16380")

# pass the content to BeautifulSoup
soup = BeautifulSoup(response.content, 'html.parser')

print(soup)
```
