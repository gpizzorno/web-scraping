```python
import requests
import json
import datetime
import time
from selenium import webdriver
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver import ActionChains
from selenium.webdriver.common.keys import Keys
import pandas as pd
import lxml.etree as et
from bs4 import BeautifulSoup
import html5lib
import csv
```

```python
# record script start time
sstime = time.time()
# define paths and urls
landing_url = 'https://archaeologydataservice.ac.uk/archives/view/greylit/query.cfm'
results_url = 'https://archaeologydataservice.ac.uk/archives/view/greylit/query.cfm?REDSQUIDARCHIVES_6ec14040-fdb2-48ae-98f0-570317ba4af9_0'
driver_path = 'chromedriver'  # update this to your local chromedriver path, or leave as-is if chromedriver is on your PATH
# define HTML parser
htmlparser = et.HTMLParser()
# create container for scraped results
output = []
# create log
results_log = []
# load selenium driver
driver = webdriver.Chrome(driver_path)
# get query landing page
driver.get('https://archaeologydataservice.ac.uk/archives/view/greylit/query.cfm')
```

```python
# click on "search" button to see all results
driver.find_element_by_name('search').click()
```

```python
# pageSource = driver.page_source
# set up loop
pageSource = driver.execute_script("return document.body.innerHTML;")
links = []
is_last = False

# only activate the following line if testing, and deactivate the line after that
# for i in range(10):
    
while is_last == False:
    # pass the content to BeautifulSoup
    soup = BeautifulSoup(pageSource, 'html.parser')
    
    # scrape links
    for link in soup.table.findAll('a'):
        links.append(link.get('href'))
    
    # attempt to go to next page
    try:
        driver.find_element_by_xpath('//*[@id="archive"]/form/div[2]/a[@title="Go forward 20 records"]').click()
        pageSource = driver.execute_script("return document.body.innerHTML;")

    # end loop if error is returned
    except:
        is_last = True
        print("Finished.")
```

```python
# create text file with list of links

filename = '../output/ADSScraping/greylitlinks.txt'
output = list(set(links))

with open(filename, 'w') as file:
    for link in output:
        file.write(link + '\n')
```
